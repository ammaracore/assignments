```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("student_tea_snacks.csv")
```

```python
data.info()
```

# q1

```python
tea = data["tea_type"].mode().iloc[0]
print("most preferred tea type:", tea)
```

# q2

```python
avg_cups = data.groupby("gender")["cups_per_day"].mean().round(2)
print("Average cup per female:",avg_cups["Female"],"cups")
print("Average cup per male:",avg_cups["Male"],"cups")
```

# q3

```python
counts = data["snack_choice"].value_counts()
plt.bar(counts.index, counts.values, color='pink', width=0.5)
plt.xlabel('Snack choice')
plt.ylabel('Number of people')
plt.title('Snack Choices')
plt.show()
```

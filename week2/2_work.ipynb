```python
clean = raw.copy()
```

```python
clean["price"] = pd.to_numeric(clean["price"], errors="coerce")
```

```python
clean["date"] = pd.to_datetime(clean["date"], errors="coerce")
```

```python
print(clean.dtypes)
```

```python
print(raw.isna().sum())
print(clean.isna().sum())
```

```python
print(clean[clean["price"].isna()])
```

```python
clean["price_missing"] = clean["price"].isna()
```

```python
print(clean["price_missing"].sum())
```

```python
#before-and-after table

# rows
print(len(raw), len(clean))

# missing counts
print(raw.isna().sum())
print(clean.isna().sum())

# date before vs after
print(raw["date"].isna().sum(), clean["date"].isna().sum())

# which rows were dropped, if any
print(raw.loc[raw.index.difference(clean.index)])

# which row got filled and with what
print(clean[clean["price_missing"]])

# item counts before vs after
print(pd.concat([raw["item"].value_counts(), clean["item"].value_counts()],
                axis=1, keys=["raw", "clean"]))
```

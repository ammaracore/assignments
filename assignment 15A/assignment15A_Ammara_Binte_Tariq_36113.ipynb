```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("ride_hailing_dataset.csv")
```

```python
data.info()
```

# q1

```python
avg_fare = data.groupby("service_type")["fare"].mean().round(2)
print(avg_fare)
```

# q2

```python
trips = data["city"].mode().iloc[0]
print("city with most trips:", trips)
```

# q3

```python
avg = (data["distance_km"].mean().round(2))
print("average distance per trip:", avg, "km")
```

# q4


```python
ratings = data.groupby("gender")["trip_rating"].mean().round(2)
if ratings["Male"] > ratings["Female"]:
  print("On average males give a higher trip rating")
else:
  print("On average females give a higher trip rating")
print(ratings)
```

# q5

```python
plt.bar(avg_fare.index,avg_fare.values, color='pink', width=0.5)
plt.xlabel('service type')
plt.ylabel('average fare (rupees)')
plt.title('Bar Chart')
plt.show()
```

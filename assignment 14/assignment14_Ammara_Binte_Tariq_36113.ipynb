```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# q5

```python
avg_by_city = data1.groupby("city")["order_value"].mean().round(2)
print(avg_by_city)
```

```python
plt.bar(avg_by_city.index,avg_by_city.values, color='pink', width=0.5)
plt.xlabel('city')
plt.ylabel('average order value')
plt.title('Bar Chart')
plt.show()
```

# DATASET 2 (student_lifestyle_dataset.csv)

```python
data2 = pd.read_csv("student_lifestyle_dataset.csv")
```

```python
data2.info()
```

# q1

```python
avg = data2["study_hours"].mean().round(2)
print("average no of study hours:", avg, "hours")
```

# q2

```python
sleep = data2.groupby("gender")["sleep_hours"].mean().round(2)
print(sleep)
if sleep["Male"] > sleep["Female"]:
  print("Males sleep more on average")
else:
  print("Females sleep more on average")
```

# q3

```python
sns.regplot(data=data2, x='screen_time', y='sleep_hours')
plt.show()
r = data2["screen_time"].corr(data2["sleep_hours"])
print(f"Correlation: {r:.2f}")
```

the regression line is nearly flat, and correlation between sleep hours and screen time is 0.05, really close to 0, hence there is no relationship between the two variables.

# q4

```python
plt.hist(data2["daily_transport_cost"], bins=15, color='pink', edgecolor='white', alpha=0.7)
plt.xlabel('Daily transport cost (pkr)')
plt.ylabel('No of students')
plt.title('Histogram')
plt.tight_layout()
plt.show()
```

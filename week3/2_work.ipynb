```python
import pandas as pd
```

```python
data = pd.read_csv(r"teaching_prices_clean.csv")
```

```python
data.info()
```

```python
data.groupby("item")["price"].agg(["mean","median"]).round(2)
```

```python
for item, group in data.groupby("item"):
    group = group.sort_values("date")

    first = group.iloc[0]["price"]
    last = group.iloc[-1]["price"]

    change = ((last - first) / first) * 100

    print(item, ":" , round(change, 2), "%")
```

```python
messy = pd.read_csv("teaching_prices_messy.csv")
```

```python
messy["price"] = pd.to_numeric(messy["price"].astype(str).str.replace(",", ""),errors="coerce")

staple = messy[messy["item"] == "Staple A"]["price"].dropna()

# With flagged value
print("With flagged value:")
print("Mean:", staple.mean().round(2))
print("Median:", staple.median())

# Without flagged value
staple_clean = staple[staple != 1250]

print("Without flagged value:")
print("Mean:", staple_clean.mean().round(2))
print("Median:", staple_clean.median())
```
